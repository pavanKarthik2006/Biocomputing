```python
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC

def compute_score(X_train,y_train,X_test,y_test):

  model=SVC(kernel='rbf',C=20,gamma='auto',random_state=42)

  model.fit(X_train,y_train)

  score=model.score(X_test,y_test)
  return score

iris=datasets.load_iris()
X=iris.data
y=iris.target

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)

score=compute_score(X_train,y_train,X_test,y_test)
print(score)
```

```python
from sklearn import datasets
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.metrics import precision_score,f1_score

iris=datasets.load_iris()
X=iris.data
y=iris.target

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)

pipe=Pipeline([
    ('Scaler',MinMaxScaler()),
    ('classifier',SVC(kernel='linear',decision_function_shape='ovr',C=1,class_weight=None))
])

pipe.fit(X_train,y_train)
y_pred=pipe.predict(X_test)

precision=precision_score(y_test,y_pred,average='weighted')
print(precision)
f1=f1_score(y_test,y_pred,average='weighted')
print(f1)
```

```python
from sklearn import datasets
import pandas as pd

# Load iris dataset
iris = datasets.load_iris()
X = pd.DataFrame(iris.data, columns=iris.feature_names)
y = pd.Series(iris.target)
target_names = iris.target_names


# Convert target to class labels
y_labels = pd.Series([target_names[i] for i in y])

# Combine into a DataFrame for filtering
iris_df = X.copy()
iris_df['target'] = y_labels

# Drop rows where class = 'Iris-versicolor'
filtered_df = iris_df[iris_df['target'] != 'Iris-versicolor']

# Separate features and target again
X_filtered = filtered_df.drop('target', axis=1)
y_filtered = filtered_df['target']
```

```python
from sklearn.preprocessing import LabelEncoder

label_enc=LabelEncoder()
y_filtered_enc=label_enc.fit_transform(y_filtered)

from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X_filtered,y_filtered_enc,test_size=0.3,random_state=0)


from sklearn.pipeline import Pipeline
from sklearn.preprocessing import MinMaxScaler
from sklearn.svm import SVC

pipeline=Pipeline([
    ('Scaler',MinMaxScaler()),
    ('classifier',SVC())
])

pipeline.fit(X_train,y_train)


from sklearn.metrics import recall_score

y_pred=pipeline.predict(X_test)
recall=recall_score(y_test,y_pred,average='weighted')
print(recall)
```

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score

def compute_score(X_train,X_test,y_train,y_test):

  model=LinearSVC(loss='hinge',penalty='l2',random_state=42,dual=True)

  model.fit(X_train,y_train)

  y_pred=model.predict(X_test)

  return accuracy_score(y_pred,y_test)

iris=load_iris()
X=iris.data
y=iris.target

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)

score=compute_score(X_train,X_test,y_train,y_test)
print(score)
```

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split,cross_val_score
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
import numpy as np

def hyperparameter_search(kernel,C_value):
  iris=load_iris()
  X=iris.data
  y=iris.target

  X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)

  model=SVC(kernel=kernel,C=C_value,random_state=42)

  scores=cross_val_score(model,X_train,y_train,cv=10)

  avg_scores=np.mean(scores)

  return avg_scores

kernels = ['linear', 'rbf']
C_values = [5, 10, 100]

for kernel in kernels:
    for C in C_values:
        avg = hyperparameter_search(kernel, C)
        print(f"Kernel: {kernel}, C: {C} --> Avg CV Score: {avg:.4f}")
```
