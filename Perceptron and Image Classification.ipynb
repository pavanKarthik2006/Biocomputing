```python
from sklearn.datasets import fetch_openml
import numpy as np
mnist = fetch_openml(name="mnist_784", version=1, as_frame=False)


num_samples = mnist.data.shape[0]
print(num_samples)

sample=mnist.data[2022]
label=mnist.target[2022]

label=int(label)

mean_value=np.mean(sample)

num_zeroes=np.sum(sample==0)

result=(label,mean_value,num_zeroes)
print(result)

```

```python
from sklearn.datasets import fetch_openml
import numpy as np

mnist=fetch_openml(name="mnist_784",version=1,as_frame=False)
X,y=mnist.data,mnist.target.astype(int)

X_train,y_train=X[:10000],y[:10000]
X_test,y_test=X[10000:12000],y[10000:12000]


unqiue_train=np.unique(y_train)

num_6=np.sum(y_train==6)
num_9=np.sum(y_train==9)

unique_test=len(np.unique(y_test))

counts_test=np.bincount(y_test)
most_common_test_class=np.argmax(counts_test)

result=(unqiue_train,num_6,num_9,unique_test,most_common_test_class)
print(result)
```

```python
X6=X_train[y_train==6]
X9=X_train[y_train==9]

y6=np.ones(len(X6))
y9=-1*np.ones(len(X9))

X_new = np.vstack((X6, X9))
y_new = np.concatenate((y6, y9))
```

```python
from sklearn.utils import shuffle
X_new,y_new=shuffle(X_new,y_new,random_state=1729)

print(y_new[:3])
```

```python
from sklearn.linear_model import Perceptron

clf=Perceptron(random_state=1729,eta0=1,max_iter=10,shuffle=False,fit_intercept=True,penalty=None,tol=None)

clf.fit(X_new,y_new)

print(clf.coef_[0][68])
```

```python
intercepts=[]
for epoch in range(5):
  clf.partial_fit(X_new,y_new,classes=[-1,1])
  intercepts.append(clf.intercept_[0])

print(intercepts)
```
