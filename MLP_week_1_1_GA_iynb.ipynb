```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('/Week1_GA_dataset (2).csv')
```

```python
unknown_count = (df=='?').sum().sum()
print(unknown_count)

```

```python
value=df.iloc[692]
print(value)
```

```python
ans=df.iloc[546,7]
print(ans)
```

```python
unique_localities=df['Locality'].unique()
print(unique_localities)
```

```python
df.replace('?',np.nan,inplace=True)
missing_values=df.isnull().sum()

missing_columns=missing_values[missing_values>0]
print(missing_columns)
```

```python
df_cleaned = df[df.isnull().sum(axis=1) <= 2]

# Step 3: Get the number of remaining samples
remaining_samples = df_cleaned.shape[0]
print(remaining_samples)
```
