```python
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.ensemble import AdaBoostClassifier

wine=load_wine()
X=wine.data
y=wine.target

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=1)

model=AdaBoostClassifier(random_state=1)
model.fit(X_train,y_train)

train_score=model.score(X_train,y_train)
test_score=model.score(X_test,y_test)

print(train_score)
print(test_score)
```

```python
from sklearn.datasets import load_wine
from sklearn.ensemble import AdaBoostClassifier
from sklearn.model_selection import train_test_split,GridSearchCV

wine=load_wine()
X=wine.data
y=wine.target

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=1)

param_grid={
    'n_estimators':[100,500,1000],
    'learning_rate':[0.5,1,2]

}

base_model = AdaBoostClassifier(random_state=1)

# Apply GridSearchCV
grid_search = GridSearchCV(estimator=base_model, param_grid=param_grid, cv=4, n_jobs=-1)
grid_search.fit(X_train, y_train)

# Best parameters and estimator
best_model = grid_search.best_estimator_

# Compute and print scores
train_score = best_model.score(X_train, y_train)
test_score = best_model.score(X_test, y_test)

print("Best Parameters:", grid_search.best_params_)
print("Training Score:", train_score)
print("Testing Score:", test_score)
```

```python
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.ensemble import BaggingClassifier, RandomForestClassifier, GradientBoostingClassifier, VotingClassifier

# Load dataset
wine = load_wine()
X, y = wine.data, wine.target

# Split the dataset
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=1)

# Define the base classifiers
clf1 = BaggingClassifier(random_state=1)
clf2 = RandomForestClassifier(random_state=1)
clf3 = GradientBoostingClassifier(random_state=1)

# Create the voting classifier
voting_clf = VotingClassifier(estimators=[
    ('bagging', clf1),
    ('rf', clf2),
    ('gb', clf3)
], voting='hard')

# Fit the model
voting_clf.fit(X_train, y_train)

# Calculate scores
train_score = voting_clf.score(X_train, y_train)
test_score = voting_clf.score(X_test, y_test)

# Print scores rounded to 4 decimal places
print("Training Score:", round(train_score, 4))
print("Testing Score:", round(test_score, 4))
```
